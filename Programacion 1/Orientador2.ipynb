## 2. Creación del módulo `videojuegos.py`
Trabajen con **dos archivos en la misma carpeta**: el módulo contiene las funciones y `principal.py` maneja la interacción con el usuario.

**Contrato de cada función:**
| Función | Comportamiento |
|---|---|
| `mostrar_juegos(juegos)` | Solo muestra la lista (sin retornar nada) |
| `buscar_juego(juegos, titulo)` | Retorna la posición o `-1` si no existe |
| `agregar_juego(juegos, titulo)` | Modifica la lista; retorna `True` o `False` |
| `cantidad_juegos(juegos)` | Retorna un número entero |

> La celda usa `%%writefile` para guardar el módulo directamente en disco.

````python
%%writefile videojuegos.py
# 2. Creación del módulo

def mostrar_juegos(juegos):
    """Solamente muestra la lista."""
    for juego in juegos:
        print("-", juego)

def buscar_juego(juegos, titulo):
    """Retorna la posición o -1."""
    if titulo in juegos:
        return juegos.index(titulo)
    return -1

def agregar_juego(juegos, titulo):
    """Modifica la lista y retorna True o False."""
    if titulo not in juegos:
        juegos.append(titulo)
        return True
    return False

def cantidad_juegos(juegos):
    """Retorna un número entero."""
    return len(juegos)
````

## 4. Rebanadas (slicing) para obtener nuevas listas
Sintaxis: `lista[inicio:fin:paso]` — incluye `inicio`, excluye `fin`. Del lado derecho de una asignación genera una **nueva lista**.

```python
juegos = ["Minecraft", "Valorant", "Fortnite", "FIFA",
          "Rocket League", "Roblox", "LOL", "Among Us"]
```

Obtengan usando **únicamente slicing**:
- **a)** Los tres primeros elementos.
- **b)** Los cuatro últimos elementos.
- **c)** Elementos desde la posición 1 hasta la 4 inclusive.
- **d)** Elementos de posiciones pares.
- **e)** Elementos de posiciones impares.
- **f)** La lista invertida.
- **g)** Todos excepto el primero.
- **h)** Todos excepto el último.

````python
# 4. Rebanadas (slicing)
juegos = ["Minecraft", "Valorant", "Fortnite", "FIFA", "Rocket League", "Roblox", "LOL", "Among Us"]

print("a)", juegos[:3])
print("b)", juegos[-4:])
print("c)", juegos[1:5])
print("d)", juegos[::2])
print("e)", juegos[1::2])
print("f)", juegos[::-1])
print("g)", juegos[1:])
print("h)", juegos[:-1])
````

## 5. Modificación de listas mediante slicing
Cuando la rebanada aparece al **lado izquierdo** de una asignación permite reemplazar, eliminar o insertar elementos en la lista **original**. Partan de una copia nueva en cada inciso.

```python
numeros = [2, 4, 6, 8, 10, 12, 14]
```

- **a)** Reemplazar `6` y `8` por `60` y `80`.
- **b)** Eliminar `10` y `12`.
- **c)** Insertar `100` y `200` entre `4` y `6` usando una **rebanada nula** (`lista[i:i]`).
- **d)** Agregar tres valores al comienzo.
- **e)** Vaciar la lista usando una rebanada.

````python
# 5. Modificación de listas mediante slicing
numeros = [2, 4, 6, 8, 10, 12, 14]

lista_a = numeros.copy()
lista_a[2:4] = [60, 80]
print("a)", lista_a)

lista_b = numeros.copy()
lista_b[4:6] = []
print("b)", lista_b)

lista_c = numeros.copy()
lista_c[2:2] = [100, 200]
print("c)", lista_c)

lista_d = numeros.copy()
lista_d[:0] = [-3, -2, -1]
print("d)", lista_d)

lista_e = numeros.copy()
lista_e[:] = []
print("e)", lista_e)
````

## 6. Listas por comprensión
Construyan **nuevas listas** sin usar `append()`. Distingan entre transformación, filtro y expresión condicional.

```python
puntajes = [45, 80, 63, 91, 100, 72]
```

| Estructura | Uso |
|---|---|
| `[expresion for e in seq]` | Transformación |
| `[expresion for e in seq if cond]` | Filtro |
| `[val_si if cond else val_no for e in seq]` | Condicional |

- **a)** Lista con todos los puntajes duplicados.
- **b)** Lista con solo los puntajes aprobados (≥ 60).
- **c)** Lista con el cuadrado de los valores pares.
- **d)** Lista con `"Aprobado"` o `"Revisar"` según cada puntaje.
- **e)** Explicar por qué la lista original **no se modifica**.

````python
# 6. Listas por comprensión
puntajes = [45, 80, 63, 91, 100, 72]

print("a)", [p * 2 for p in puntajes])
print("b)", [p for p in puntajes if p >= 60])
print("c)", [p**2 for p in puntajes if p % 2 == 0])
print("d)", ["Aprobado" if p >= 60 else "Revisar" for p in puntajes])
print("e) La lista original no se modifica porque la sintaxis por comprensión genera una lista completamente nueva en memoria.")
````

## 7. Funciones lambda
Una función `lambda` expresa una **operación breve** en una sola línea. No reemplaza a `def` cuando se necesitan varias instrucciones, validaciones complejas o documentación.

```python
nombre = lambda param: expresion
```

Definan y prueben:
- **a)** `cuadrado` — retorna el cuadrado de un número.
- **b)** `es_par` — retorna `True` si el número es par.
- **c)** `mayor` — retorna el mayor entre dos números.
- **d)** `aplicar_descuento` — aplica un descuento porcentual a un precio.
- **e)** ¿Cuál escribirían con `def` en un programa real? Justifiquen.

````python
# 7. Funciones lambda
cuadrado = lambda x: x**2
es_par = lambda x: x % 2 == 0
mayor = lambda a, b: a if a > b else b
aplicar_descuento = lambda precio, desc: precio - (precio * (desc / 100))

print("a)", cuadrado(5))
print("b)", es_par(4))
print("c)", mayor(10, 20))
print("d)", aplicar_descuento(1000, 20))
print("e) Escribiría 'aplicar_descuento' con 'def' en la vida real si necesitara verificar que el precio y el descuento no sean valores negativos.")
````

## 8. Desafío integrador — Sistema de gestión de videojuegos
Desarrollen un sistema dividido en **dos módulos**: `gestion_videojuegos.py` (operaciones del catálogo) y la celda principal (entrada/salida).

El programa debe:
- Crear una lista inicial con **10 títulos**.
- Mostrar el catálogo y su cantidad de elementos.
- Buscar un título y agregarlo solo si no existe.
- Mostrar primeros 5, últimos 3 y catálogo invertido con **slicing**.
- Crear por **comprensión** una lista de títulos con más de 8 caracteres.
- Usar al menos una **función lambda**.
- Mantener en el módulo todas las operaciones del catálogo.

````python
%%writefile gestion_videojuegos.py
# 8. Módulo del Desafío Integrador

def crear_catalogo_inicial():
    return ["Halo", "Minecraft", "Zelda", "Metroid", "Mario", "Fortnite", "Valorant", "Hades", "Celeste", "Portal"]

def mostrar_catalogo(lista):
    print(f"Catálogo ({len(lista)} elementos):")
    for j in lista:
        print(f" - {j}")

def buscar_titulo(lista, titulo):
    return lista.index(titulo) if titulo in lista else -1

def agregar_titulo(lista, titulo):
    if titulo not in lista:
        lista.append(titulo)
        return True
    return False

def aplicar_filtro_longitud(lista):
    return [j for j in lista if len(j) > 8]

es_titulo_largo = lambda titulo: len(titulo) > 8
````

| **e)** | Rebanada cuyo índice final **supera** la longitud |

````python
# 8 y 9. Principal del Desafío Integrador y Verificación
import gestion_videojuegos as gv

catalogo = gv.crear_catalogo_inicial()

print("--- 9. Casos de Prueba ---")
print("a) Búsqueda existente ('Zelda'): Posición", gv.buscar_titulo(catalogo, "Zelda"))
print("b) Búsqueda inexistente ('Fifa'): Posición", gv.buscar_titulo(catalogo, "Fifa"))
print("c) Agregar repetido ('Minecraft'):", gv.agregar_titulo(catalogo, "Minecraft"))
gv.agregar_titulo(catalogo, "Terraria") # Agregar uno nuevo

print("\n--- 8. Requerimientos de Slicing ---")
print("Primeros 5:", catalogo[:5])
print("Últimos 3:", catalogo[-3:])
print("Invertido:", catalogo[::-1])

print("\n--- 9. Casos Especiales ---")
print("d) Lista vacía:")
gv.mostrar_catalogo([])
print("e) Rebanada excedida (0:50):", catalogo[0:50])

print("\n--- 8. Comprensión y Lambda ---")
juegos_largos = gv.aplicar_filtro_longitud(catalogo)
print("Juegos con nombre largo (>8):", juegos_largos)
print("Prueba Lambda 'Minecraft' largo:", gv.es_titulo_largo("Minecraft"))
````
