### 📖 Problema: Multas de Biblioteca

> Una biblioteca universitaria cobra una multa de **$800** por cada día de atraso.
> Si el libro fue devuelto con al menos un día de atraso y estaba **reservado** por otra persona,
> se agrega un cargo fijo de **$2500**. La cantidad de días no puede ser negativa.

---


**a.** 🔍 Identifiquen los **datos de entrada**, el **procesamiento** y la **salida**.

**b.** 🧪 Propongan al menos **cuatro casos de prueba**, incluyendo situaciones límite.

**c.** 📝 Escriban el **algoritmo** en lenguaje natural o pseudocódigo.

**d.** 🔄 Representen la solución mediante un **diagrama de flujo**.

**e.** 🐍 Implementen el **programa en Python** y comparen los resultados obtenidos con los casos de prueba.


```python
""" ENTRADA:
- Días de atraso
- ¿Libro reservado por otra persona? S/N

PROCESAMIENTO:
La multa es = $800 × días de atraso
- Si días > 0 AND reservado == S: agregar $2500
- Validar que días >= 0

SALIDA:
- Monto total de multa ($)
"""
```

```python
"""
============================================================
b) CASOS DE PRUEBA
============================================================
Caso 1: días=0, reservado=False, esperado=$0
Caso 2: días=5, reservado=False, esperado=$4000
Caso 3: días=3, reservado=True, esperado=$4900
Caso 4: días=0, reservado=True, esperado=$0 
"""
```

```python
""" ===========================================================
c) ALGORITMO EN PSEUDOCÓDIGO
============================================================

FUNCIÓN calcular_multa(días, reservado)
    SI días < 0 ENTONCES
        RETORNAR error
    FIN SI

    multa ← días × 800

    SI días > 0 Y reservado == VERDADERO ENTONCES
        multa ← multa + 2500
    FIN SI

    RETORNAR multa
FIN FUNCIÓN """
```

```python
# d) Función principal
def validar_datos(dias):
    if dias < 0:
        print("Error: ingrese un número CORRECTO (>= 0)")
        return        
    return dias       


def calcular_multa(dias, reservado):
    multa = dias * 800
    if dias > 0 and reservado == "S":
        multa += 2500

    return multa


# Programa principal
dias = int(input("Ingrese los días de atraso: "))
reservado = input("¿El libro estaba reservado? (S/N): ")
resultado = calcular_multa(dias, reservado)

print(resultado)
```
